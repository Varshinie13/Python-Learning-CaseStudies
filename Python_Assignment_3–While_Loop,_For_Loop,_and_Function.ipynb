Task 1: While Loop & Control Statements -Number Guessing Game

```python
import random

# 1. Set Up the Game
# Generate random number between 1 and 10
secret_number = random.randint(1, 10)

# 2. Prompt the User
print("Welcome to the Number Guessing Game!")
attempts = 3

# 3. Implement the Guessing Logic
while attempts > 0:
    # Take input from the user
    try:
        guess = int(input("Guess the number (between 1 and 10): "))
    except ValueError:
        print("Invalid input. Please enter a valid number.")
        continue

    # Check if the guess is out of the valid range
    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        # 4. Control Statements: Use continue to skip the rest of the loop
        continue

    # Decrement the attempts left only after a valid guess within range
    attempts -= 1

    # Check the guess against the secret number
    if guess == secret_number:
        print("Congratulations! You guessed the correct number.")
        # 4. Control Statements: Use break to exit the loop
        break
    elif guess > secret_number:
        if attempts > 0:
            print("Too high. Try again.")
    else:
        if attempts > 0:
            print("Too low. Try again.")

# 4. Control Statements: Use else with the while loop
else:
    print(f"\nGame Over! You ran out of attempts.")
    print(f"The correct number was: {secret_number}")
    print("Better luck next time!")
```

Task 2:- For Loop - Multiplication Table Generator

```python
# 1. Prompt user for input
# We convert the input to an integer so we can perform math operations on it
try:
    number = int(input("Enter the number for which you want the multiplication table: "))

    print(f"\nMultiplication Table for {number}:")
    print("-" * 20)

    # 2. Generate the Multiplication Table
    for i in range(1, 11):

        result = number * i

        # 3. Display the Multiplication Table
        print(f"{number} x {i} = {result}")

except ValueError:
    print("Please enter a valid whole number.")
```

Task 3: Function- BMI Calculator

```python
# 1. Define a function calculate_bmi(weight, height) that returns the BMI
def calculate_bmi(weight, height):

    bmi = weight / (height ** 2)
    return bmi

# Main program flow
try:
    print("--- Body Mass Index (BMI) Calculator ---")

    # 2. Prompt the user for their weight (in kg) and height (in meters)
    weight = float(input("Enter your weight in kilograms (e.g., 70): "))
    height = float(input("Enter your height in meters (e.g., 1.75): "))


    if weight <= 0 or height <= 0:
        print("Weight and height must be positive numbers.")
    else:
        # 3. Use the function to calculate and display the BMI
        user_bmi = calculate_bmi(weight, height)

        print(f"\nYour calculated BMI is: {user_bmi:.2f}")

        print("BMI Categories:")
        print("- Underweight: < 18.5")
        print("- Normal weight: 18.5 – 24.9")
        print("- Overweight: 25 – 29.9")
        print("- Obesity: 30 or greater")

except ValueError:
    print("Invalid input. Please enter numbers only.")
```
