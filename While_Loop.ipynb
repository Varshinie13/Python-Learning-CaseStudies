print numbers from 1 to 25 using While Loop:

```python
i = 1

while i<=25:
  print(i)
  i+=1
print("The Block Executed")
```

print Even Number between 1 to 100


```python
x = 1

while x<=100:
  x+= 1
  if x%2 == 0:
    print(x)
    continue
print("The Block Executed")
```

print odd numbers from 1 to 25

```python
n = 1

while n<=100:
  n+= 1
  if n%2 != 0:
    print(n)
    continue
print("The Block Executed")
```

print the Name of the students in the list using while loop.


```python
Students = ["Anitha","Vaishu","Vikram","Arjun","Karthick","Ajith","Srinithi"]
i = 0
while i<len(Students):
  print(Students[i])
  i+=1
```

Case Study - Password Checking

```python
password = "python123"
user_password = input("Enter the password: ")

while user_password != password:
  print("Incorrect password! , Try again")
  user_password = input("Enter the password: ")

else:
  print("Access Granted!")
```

searching a value in the list using break:

```python
num = [1,2,3,4,5,6,7,8,15,9,10]

i = 0

while i<len(num):
  print(num[i])

  if num[i] == 15:
    print("Found the number 15")
    break
  i+=1
```

Case Study - password Verification System

```python
correct_pin = 5678

while True:
  user_pin = int(input("Enter the password: "))
  if user_pin == correct_pin:
    print("Login Successfull!")
    break
  else:
    print("Incorrect Password! , Try again")
    user_pin = int(input("Enter the password: "))

```

Case Study – Password Verification System with Limited Attempts

```python
crt_password = "python123"

attempts = 0
max_attempts = 3

while attempts < max_attempts:
    user_input = input("Enter password: ")
    if user_input == crt_password:
        print("Access granted!")
        break

    else:
        attempts += 1
        print(f"Incorrect password, try again. ({max_attempts - attempts} attempts left)")

if attempts == max_attempts:
    print("Maximum attempts reached. Access denied!")
```

Continue Statement - Case Study - Student Marks Filter System

```python
students = ["Srija", "Vinitha", "Arjun", "Deva", "Elizabeth","Vishnu","Karthick"]
marks = [35,88,42,76,39,55,63]

i = 0

while i < len(students):
    if marks[i] >= 50:
        i += 1
        continue

    print(f"{students[i]} scored {marks[i]} (Below 50)")

    i += 1
```
