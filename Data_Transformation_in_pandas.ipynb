Data Transaformation

```python
import pandas as pd
```

```python
# Create first DataFrame (Employee details)
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4, 5, 6],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Age': [25, 30, 28, 35, 26, 40],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'Marketing', 'Finance'],
    'Salary': [45000, 60000, 52000, 75000, 48000, 82000]
})

# Create second DataFrame (Location details)
df2 = pd.DataFrame({
    'ID': [1, 2, 3, 4, 6, 7],
    'City': ['Delhi', 'Mumbai', 'Bangalore', 'Chennai', 'Hyderabad', 'Pune'],
    'State': ['Delhi', 'Maharashtra', 'Karnataka', 'Tamil Nadu', 'Telangana', 'Maharashtra'],
    'Country': ['India', 'India', 'India', 'India', 'India', 'India']
})
print(df1)
print(df2)
```

Merge

```python
#INNER JOIN (only matching IDs)
df_inner = pd.merge(
    df1[['ID', 'Name', 'Salary']],
    df2[['ID', 'City']],
    on='ID',
    how='inner'
)
df_inner
```

```python
#LEFT JOIN (all rows from df1)
df_left = pd.merge(
    df1,
    df2,
    on='ID',
    how='left'
)
df_left
```

```python
# RIGHT JOIN (all rows from df2)
df_right = pd.merge(
    df1[['ID', 'Name', 'Salary']],
    df2[['ID', 'City']],
    on='ID',
    how='right'
)
df_right
```

```python
# FULL OUTER JOIN (all rows from both tables)
df_outer = pd.merge(
    df1,
    df2,
    on='ID',
    how='outer'
)
df_outer
```

Joins

```python
# Set ID as index in both
print(df1)
print(df2)

df1 = df1.set_index('ID')
df2 = df2.set_index('ID')

print(df1)
print(df2)
```

```python
#INNER JOIN (common IDs only)
df_inner = df1.join(df2, how='inner')
print(df_inner)
```

```python
# LEFT JOIN (all rows from df1)
df_left = df1.join(df2, how='left')
print(df_left)
```

```python
# RIGHT JOIN (all rows from df2)
df_right = df1.join(df2, how='right')
print(df_right)
```

```python
# FULL OUTER JOIN (all rows from both)
df_outer = df1.join(df2, how='outer')
print(df_outer)
```

```python
#Reset index for both DataFrames\
df1 = df1.reset_index()
df2 = df2.reset_index()
```

```python
print(df1)
print(df2)
```

Concat

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'ID': [4, 5, 6],
    'Name': ['David', 'Eva', 'Frank']
})
```

```python
result = pd.concat([df1, df2],axis=0,ignore_index=True)
print(result)
```

```python
result_col = pd.concat([df1, df2], axis=1)
print(result_col)
```

```python
df3 = pd.DataFrame({
    'ID': [7, 8],
    'City': ['Delhi', 'Mumbai']
})

result_mixed = pd.concat([df1, df3],ignore_index=True)
print(result_mixed)
```

Groupby

```python
#sample data
data = {
    'Region': ['East', 'West', 'East', 'North', 'West', 'East'],
    'Sales': [200, 300, 250, 400, 150, 350],
    'Profit': [20, 30, 25, 40, 15, 35]
}

df =pd.DataFrame(data)
```

```python
# 1. Summarize Total Sales by Region
# Find total sales generated in each region

print(df.groupby("Region")["Sales"].sum())
```

```python
# 2. Find Average Profit by Region
# Calculate average profit for each region

print(df.groupby("Region")["Profit"].mean())
```

```python
# 3. Count Number of Records by Region
# Count how many sales records exist in each region

print(df.groupby("Region")["Sales"].count())
region_count =df.groupby("Region")["Sales"].count()
region_count
```

```python
# 4. Find Maximum Sales by Region
# Identify highest an lowest sales and profit value in each region

min_max = df.groupby("Region")[["Sales","Profit"]].agg(["min","max"])
min_max
```

Multiple Aggregations :.agg() if you want to apply one or more aggregation in column

```python
# 5. Perform Multiple Aggregations
# Generate multiple summary statistics together
df.groupby("Region")[["Sales", "Profit"]].agg(["sum", "mean", "max", "min"])
```

Cross Tab in Pandas

```python
# Sample Data
data = {
    'Region': ['East', 'West', 'East', 'North', 'West', 'East'],
    'Category': ['Electronics', 'Furniture', 'Electronics',
                 'Furniture', 'Clothing', 'Clothing'],
    'Sales': [200, 300, 250, 400, 150, 350],
    'Profit': [20, 30, 25, 40, 15, 35]
}

df = pd.DataFrame(data)
```

```python
# Objective:Analyze frequency distribution between Region and Category
cross_tab = pd.crosstab(df["Region"],df["Category"])
cross_tab
```

```python
cross_tab_multi_sum = pd.crosstab(df["Region"],df["Category"],
                              values=df["Sales"],aggfunc=["sum"]
)
cross_tab_multi_sum
```

```python
# CROSS TAB WITH MULTIPLE AGGREGATIONS
# Objective: Calculate sum and average sales
cross_tab_multi = pd.crosstab(df["Region"],df["Category"],
                              values=df["Sales"],aggfunc=["sum", "mean"]
)

cross_tab_multi
```
