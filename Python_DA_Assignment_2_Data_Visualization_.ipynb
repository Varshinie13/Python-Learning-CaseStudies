Loading the Taxis Dataset:

```python
import seaborn as sns
# Load the 'taxis' dataset
df = sns.load_dataset("taxis")
```

Data Cleaning

```python
#Loading the dataset

df
```

```python
#Loading top 5 data for analysing
df.head()
```

```python
#returns column names
df.columns
```

```python
#returns the data type of the column
df.dtypes
```

```python
#returns the null value counts

df.isnull().sum()
```

```python
#impute the missing values

Fill_cols = ['payment','pickup_zone','dropoff_zone','pickup_borough','dropoff_borough']
df[Fill_cols] = df[Fill_cols].fillna(df[Fill_cols].mode().iloc[0])
```

```python
#check again the null values

df.isnull().sum()
```

```python
#returns the count of duplicate values

df.duplicated().sum()
```

Visualizations using Matplotlib/Pandas Plot:  

```python
import matplotlib.pyplot as plt
```

```python
df.columns
```

```python
#Line Chart
plt.figure(figsize=(20,5))
df_sorted = df.sort_values(by='pickup')
plt.plot(df_sorted['pickup'],df_sorted['fare'])
plt.xlabel('Pickup Date and Time')
plt.ylabel('Fare for the location')
plt.title('Fare over time')
plt.xticks(rotation = 45)
plt.show()
```

```python
#Bar Chart
plt.figure(figsize = (10,7))
df_grouped = df.groupby('pickup_borough')['fare'].sum()
ax = df_grouped.plot(kind = 'bar',color = 'Green', edgecolor = 'Black')
ax.bar_label(ax.containers[0],padding = 3)
plt.title('Total Fare of each pickup_borough')
plt.ylabel = 'Total Fare'
plt.show()
```

```python
payment_method_grp = df.groupby('payment')['total'].count()
payment_method_grp
```

```python
#pie chart
plt.figure(figsize = (10,5))
payment_method_grp = df.groupby('payment')['total'].count().plot(kind = 'pie',autopct = '%1.1f%%',startangle = 140,ylabel = '')
plt.title('Distribution of specific payment methods')
plt.show()

```

```python
# Box Plot

plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='pickup_borough', y='tip', palette='pastel')
plt.title('Distribution of Tip Amounts by Pickup Borough', fontsize=14, pad=15)
plt.xlabel('Pickup Borough', fontsize=12)
plt.ylabel('Tip Amount', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Visualizations using Seaborn:  

```python
import seaborn as sns
```

```python
#Counter plot

plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='pickup_borough', palette='viridis', order=df['pickup_borough'].value_counts().index)
plt.title('Number of Trips in Each Pickup Borough', fontsize=14, pad=15)
plt.xlabel('Pickup Borough')
plt.ylabel('Count of Trips')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Scatter plot

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='distance', y='fare', hue='pickup_borough')
plt.title('Relationship Between Distance and Fare by Pickup Borough', fontsize=14, pad=15)
plt.xlabel('Distance')
plt.ylabel('Fare')
plt.legend(title='Pickup Borough', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
#Heatmap

plt.figure(figsize=(8, 6))
# Selecting only the numerical columns
numerical_cols = ['distance', 'fare', 'tip', 'tolls', 'total']
correlation_matrix = df[numerical_cols].corr()

# Plotting the heatmap
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5, square=True)
plt.title('Correlation Matrix of Numerical Variables', fontsize=14, pad=15)
plt.tight_layout()
plt.show()
```

```python
# Pair Plot

pairplot_cols = ['distance', 'fare', 'tip', 'total', 'pickup_zone']
g = sns.pairplot(df[pairplot_cols], hue='pickup_zone', palette='husl', plot_kws={'alpha': 0.6})
g.fig.suptitle('Pairwise Relationships by Pickup Zone', y=1.02, fontsize=14)
plt.show()
```

```python
#Violin plot
plt.figure(figsize=(10, 6))
sns.violinplot(data=df, x='payment', y='fare', palette='muted')
plt.title('Distribution of Fare for Each Payment Method', fontsize=14, pad=15)
plt.xlabel('Payment Method', fontsize=12)
plt.ylabel('Fare', fontsize=12)
plt.tight_layout()
plt.show()
```
