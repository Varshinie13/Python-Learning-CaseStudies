Case Study - Text Analyser

```python
Str = input("Enter the String: ").lower()

vowel = consonant = digit = special_char = 0

for letter in Str:
    if letter in 'aeiou':
        vowel += 1
    elif letter.isalpha():
        consonant += 1
    elif letter.isdigit():
        digit += 1
    else:
        special_char += 1

print(f"Vowels: {vowel}, Consonants: {consonant}, Digits: {digit}, Special: {special_char}")
```

Case Study - Restricted word detected

```python
Words = ["Hello","python","Welcome","To","Learning","process","Blocked","Happy","Journey"]

restricted_word = "Blocked"

for word in Words:
  if word == restricted_word:
    print("Restricted Word found")
    break
  print(word)
```

```python
#Count the Restricted words
Words = ["Hello", "python", "Welcome", "To", "Learning", "process", "Blocked", "Happy", "Journey", "Wrong", "Hacking"]
restricted_words = ["Blocked", "Wrong", "Hacking"]
count = 0


for word in Words:
    if word in restricted_words:
        print(f"-> Alert: Found restricted word: '{word}'")
        count += 1

print(f"\nTotal restricted words found: {count}")
```

Case Study -  Employee Attendance Count System using continue

```python
attendance = ["Present", "Absent", "Present", "Leave", "Present","Absent","Present","Leave"]


present = 0
absent = 0
leave = 0


for status in attendance:


    if status == "Present":
        present += 1
        continue
    if status == "Absent":
        absent += 1
        continue
    if status == "Leave":
        leave += 1
        continue


print(f"Total Present: {present}")
print(f"Total Absent:  {absent}")
print(f"Total Leave:   {leave}")
```

Case Study - Student Performance Management System Instructions


```python
# ==========================================
# CASE STUDY: STUDENT PERFORMANCE SYSTEM
# ==========================================

# 1. Creating the Lists
students = ["Alice", "Bob", "Charlie", "David", "Eva"]
marks = [85, 92, 78, 95, 88]

# ------------------------------------------
# HEADING 1: Using range()
# ------------------------------------------
print("## 1. Student Records (Accessed via range() and Indexes)")
print("-" * 55)

# range(len(...)) generates index positions (0 to 4)
for i in range(len(students)):
    print(f"Index {i} -> Student: {students[i]:<8} | Marks: {marks[i]}")

print("\n" + "="*55 + "\n")

# ------------------------------------------
# HEADING 2: Using enumerate()
# ------------------------------------------
print("## 2. Student Roster (Accessed via enumerate())")
print("-" * 55)

# enumerate() automatically gives both the index and the item
for index, name in enumerate(students, start=1):
    print(f"Roll No: {index} | Name: {name}")

print("\n" + "="*55 + "\n")

# ------------------------------------------
# HEADING 3: Using zip()
# ------------------------------------------
print("## 3. Performance Summary (Combined via zip())")
print("-" * 55)

# zip() pairs elements from both lists together into tuples
for name, score in zip(students, marks):
    print(f"Student: {name:<8} achieved a score of {score}%")

print("\n" + "="*55 + "\n")

# ------------------------------------------
# HEADING 4: Final Clean Format
# ------------------------------------------
print("## 4. Final Performance Report Card")
print("-" * 55)
print(f"{'ID':<5} | {'Student Name':<12} | {'Marks':<5} | {'Status':<8}")
print("-" * 55)

for i, (name, score) in enumerate(zip(students, marks), start=1):
    status = "Pass" if score >= 40 else "Fail"
    print(f"{i:<5} | {name:<12} | {score:<5} | {status:<8}")

print("-" * 55)
```
