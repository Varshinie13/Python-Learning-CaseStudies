Case Study - College Management System

```python
#Create a list called participants and perform add, insert, remove, and display operations.

Participants = ["Ashok","Srija","Vinitha","Banu","Krithika","Deva"]
print(f"The Participants are:",Participants)
```

```python
# add an element in the list

Participants.append("Vinoth")
print(f"The Participants are:",Participants)
```

```python
# insert an element in the list
Participants.insert(2,"Karthick")
print(f"The Participants are:",Participants)
```

```python
# remove an element in the list
Participants.remove("Karthick")
print(f"The Participants are:",Participants)
```

```python
#Display NAme of participants

print(f"Final Participants Names are" ,Participants)
```

Create a list -  departments

```python
#Create a list called departments

Departments = ["HR", "IT", "Finance", "HR", "Marketing","Sales","IT"]
print(Departments)
```

```python
#convert the list into a set to remove duplicates

Departments = {"HR", "IT", "Finance", "HR", "Marketing","Sales","IT"}
print(Departments)
print(type(Departments))
```

```python
#perform add and remove operations

Departments.add("Operations")
print(Departments)
```

```python
#Remove Department Names

Departments.remove("Operations")
print(f"Department Name:",Departments)
```
