Numpy Array Operations:  

```python
import numpy as np
```

```python
#Create a 1D NumPy Array:

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
temperatures_w1
```

```python
#Inspection and Properties:

#Inspect the shape
print(f"The shape of the array : ",temperatures_w1.shape)

#Inspect the DataType
print(f"The array data Type is: ",temperatures_w1.dtype)

#inspect Number of elements in array
print(f"The number of elements in array: ",temperatures_w1.size)
```

```python
#Array Operations:

#Convert the temperatures_w1 from Celsius to Fahrenheit :
Fahrenheit = (temperatures_w1 * 9/5) + 32
print(f"The celsius Temperatures: ",temperatures_w1)
print(f"The Fahrenheit Temperatures: ",Fahrenheit)
```

```python
# maximum temperature of the week
max_temp = np.max(temperatures_w1)
print(f"The Maximum Temparature of the week is : ",max_temp)
#minimum temperature of the week
min_temp = np.min(temperatures_w1)
print(f"The Minimum Temparature of the week is : ",min_temp)
```

```python
#mean temparature of the week
Avg_temp = np.mean(temperatures_w1)
print(f"The Average Temparature of the week is : {Avg_temp:.2f}")
```

Pandas Series:  

```python
import pandas as pd
```

```python
#Creating Pandas Series:
marks = [95, 92, 89, 85, 80 ]
indices = ["Rank1","Rank2","Rank3","Rank4","Rank5"]

scores = pd.Series(marks, index = indices)
scores
```

Indexing and Slicing:  

```python
# Use the integer index position to access the mark of the 1st rank student.

print(f"The 1st Rank student Marks is : ",scores.Rank1 )
print("------------------------------------------------------------------------")
# Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.

print(f"The Top 3 ranks:\n",scores.loc['Rank1':'Rank3'])
print("------------------------------------------------------------------------")

# Use the iloc accessor to retrieve the mark of the 3rd-ranked student.

print(f"The 3rd ranked student :",scores.iloc[2])
print("------------------------------------------------------------------------")
# Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.

print(f"Marks greater than 90:\n",scores[(scores>90)])
```

Manipulating Series:

```python
scores
```

```python
# Modify the mark of the student with the 1st rank to 100.
# Remove the entry corresponding to the last rank from the Series.
scores.iloc[0]=100
scores
scores.drop('Rank5',inplace = True)


```

```python
# Compute the CGPA by dividing each mark in the Series by 10.
cgpa = scores/10
print(f"CGPA:\n",cgpa)
```

Manipulating the DataFrame:  

```python
Transactions = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110 ],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing','Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South','North' ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400 ]}
Tansactions = pd.DataFrame(Transactions)
Tansactions
```

```python
# Modify the 'Amount' for the transaction with TransactionID 102 to 165.

Tansactions.loc[Tansactions['TransactionID'] == 102,'Amount'] = 165
```

```python
Tansactions
```

```python
#Add a new column 'Discount' by calculating 10% of the 'Amount'.

Tansactions['Discount'] = Tansactions['Amount']*0.10
Tansactions
```

```python
#Remove the row with TransactionID 109.
Tansactions = Tansactions[Tansactions['TransactionID']!= 109]
Tansactions
```

```python
#Delete the 'Discount' column from the DataFrame.
Tansactions = Tansactions.drop('Discount',axis = 1)
Tansactions
```
