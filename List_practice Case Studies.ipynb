Case Study – Student Marks Analysis
Instructions:

Create a list named marks by getting 5 subject marks from the user using map() and list()

Perform calculations:

• Find total marks using sum()

• Find average marks using len()

• Calculate percentage assuming each subject is out of 100

```python
Marks = list(map(int,input("Enter the Marks: ").split(",")))
Marks
```

```python
#Total Marks

Total_Marks = sum(Marks)
print("Total Marks Scored:",Total_Marks)
```

```python
#Avg Marks
average_Marks = Total_Marks/len(Marks)
print("Average Mark Scored:", average_Marks)
```

```python
#Percentage of marks
max_Marks = len(Marks)*100
percentage = (Total_Marks/max_Marks)*100
print("Percentage:",percentage,"%")
```

```python
print("Maximum Marks:",max(Marks))
print("Minimum Marks:",min(Marks))
```

List Comprehension

```python
mul = [num*5 for num in range(1,11)]
mul
```

```python
# Even numbers 1- 100

even_num = [x for x in range(1, 101) if x % 2 == 0]
print("Even Numbers :", even_num)

# Odd numbers 1- 100

odd_num = [x for x in range(1, 101) if x % 2 != 0]
print("Odd Numbers :", odd_num)
```

Case Study – String Analysis using List Comprehension

```python
text = "Python is a popular, high-level, interpreted programming language known for its clear syntax and code readability. "

#Convert all characters into uppercase

upper_text = [ x.upper() for x in text]
upper_text

#Extract only vowels from the string

vowels = [ x for x in text if x in "aeiou"]

#Extract only consonants from the string

consonents = [ x for x in text if x not in "aeiou"]

# Remove spaces
remove_spaces = [x for x in text if x != " "]

print("Uppercase Letters :", upper_text)
print("Vowels            :", vowels)
print("Consonants        :", consonents)
print("Without Spaces    :", remove_spaces)
```

```python
# In the given list extract the postive numbers

x = [-1,-4,6,7,8,10,-11,-15]

postive_num = [x for x in x if x > 0]
print("Postive Numbers:",postive_num)
```

```python
#From the Given list filter the sales amount greater than 1000 and sum the sales amount

Sales_amt = [1000,540,300,6000,700,200,1500,500,450,800,1900,630]

Total_amt = [x for x in Sales_amt if x > 1000]
print("Total Sales Amount:",sum(Total_amt))
```

Case Study - Employee Salary Increment System

```python
salary = [35000, 50000, 54000, 70000, 80000]

# Increase each salary by 10%
updated_salary = [i * 1.10 for i in salary]

# Display only salaries greater than 50000
high_salary = [i for i in updated_salary if i > 50000]

print("Original Salary :", salary)
print("Updated Salary  :", updated_salary)
print("High Salary     :", high_salary)
```
