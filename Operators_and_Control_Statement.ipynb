Arithematic Operators

```python
a = int(input("Enter the Number: "))
b = int(input("Enter the Number: "))

print (f"The Addition of two Numbers : ", (a+b))
print (f"The subtraction of two Numbers : ", (a-b))
print (f"The Multiplication of two Numbers : ", (a*b))
print (f"The Division of two Numbers : ", (a/b))
print (f"The Modulus of two Numbers : ", (a%b))
print (f"The Floor Division of two Numbers : ", (a//b))
print (f"The Exponentian of two Numbers : ", (a**b))
```

Comparison Operators


```python
c = int(input("Enter the Number: "))
d = int(input("Enter the Number: "))

print (f"The Given Number is Equal: ",(c == d))
print (f"The Given Number is NotEqual: ",(c != d))
print (f"The Given Number is Greater than: ",(c > d))
print (f"The Given Number is Lesser than: ",(c < d))
print (f"The Given Number is Greater or Equal: ",(c >= d))
print (f"The Given Number is Lesser or Equal: ",(c <= d))
```

Bitwise Operator

```python
a = 6 & 3
print(f"Bitwise and: {a}")

b = 6 | 3
print(f"Bitwise or: {b}")

c = 6 ^ 3
print(f"Bitwise xor: {c}")

d = ~6
print(f"Bitwise not: {d}")

e = 6 << 2
print(f"Bitwise left shift: {e}")

f = 6 >> 2
print(f"Bitwise right shift: {f}")
```

```python
# check the vote Eligibity and Citizenship

age = int(input("Enter the Age of the Person : "))
Citizenship = input("Enter the Citizenship : ")
if age >= 18:
  if Citizenship == "India":
    print("You are Indian Citizen so Eligible to Vote")
  else:
    print("You are not Indian Citizen so not eligible to Vote")
else:
  print("Not Eligible to Vote")
```

Case Study - Check Bank Loan Eligibility

```python
Customer_Name = input("Enter the Customer Name : ")
Customer_age = int(input("Enter the Age : "))
Customer_income = float(input("Enter the Income of Customer : "))
Credit_Score = int(input("Enter the Credit Score : "))
Existing_Loan_Status = input("Enter the Existing Loan Status : ").lower() == "No"

if Customer_age >= 21 and Customer_income >= 30000:
    if Credit_Score >= 700 and not Existing_Loan_Status:
        Loan_status = "Approved "
    elif Credit_Score >= 600 and Customer_income  >= 25000:
        Loan_status = "Conditionally Approved "
    else:
        Loan_status = "Rejected "
else:
    Loan_status = "Not Eligible "

print("--------Customer Details-------")

print("Customer_Name:", Customer_Name)
print("Customer_Age:", Customer_age)
print("Customer_Income:", Customer_income)
print("Credit_Score:", Credit_Score)
print("Loan_Status:", Loan_status)
```

Case Study - Scholarship Eligibility Checker

```python
Student_Name = input("Enter the Student Name : ")
Mark = int(input("Enter the Mark : "))
Attendance = input("Enter the attendence : ")
Family_Income = float(input("Enter the Family Income : "))

if Mark>= 90 and Attendance >= 90:
    Scholarship = "Full Scholarship "
elif Mark >= 80 or Attendance >= 85:
    Scholarship = "Half Scholarship "
elif Mark >= 70 and Family_Income < 30000:
    Scholarship = "Need-Based Scholarship "
else:
    Scholarship = "Not Eligible "

print( "----------Student Details--------")

print("Name of the Student:", Student_Name)
print("Student_Mark:", Mark)
print("Student_Attendance:", Attendance)
print("Student - Family Income:", Family_Income)
print("Scholarship Status of the Student:", Scholarship)
```
