```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Redefine df with the correct data for the pie chart
data = {
    'Product': ['Product A', 'Product B', 'Product C', 'Product D', 'Product E'],
    'Sales': [250, 400, 150, 300, 200],
    'Profit': [50, 80, 20, 60, 40],
    'Region': ['North', 'South', 'East', 'West', 'Central'],
    'Region_Sales': [300, 250, 400, 200, 350]
}
df = pd.DataFrame(data)
```

```python
plt.plot(df['Product'], df['Sales'], marker='+', linestyle='--', linewidth=1, color='black')
plt.show()
```

```python
#Line Chart

plt.plot(df['Product'], df['Sales'], marker='*', linestyle='--', linewidth=2, color='black')

plt.xlabel("Products", color='red', fontstyle='italic')
plt.ylabel("Total Sales", color='red', fontstyle='italic')
plt.grid(axis='y', linestyle='--')
plt.legend(['Sales'], title='Sales')

# Add data labels
for x, y in zip(df['Product'], df['Sales']):
    plt.text(x, y, str(y), ha='center', va='bottom')
```

```python
#bar chart
plt.figure(figsize=(6,4))
bars = plt.bar(df['Region'], df['Profit'], color='green')
plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.bar_label(bars)
plt.show()
```

```python
#Scatter plot

plt.scatter(df['Sales'], df['Profit'], )
plt.title("Profit vs Sales")
plt.xlabel("Sales")
plt.ylabel("Total Profit")


for x, y in zip(df['Sales'], df['Profit']):
    plt.text(x, y, f'({x},{y})', ha='center', va='bottom')

plt.show()
```

```python
#Histogram
employee_age =np.random.randint(20,50,200)
print(employee_age)

plt.figure(figsize=(6,4))
plt.hist(employee_age,bins =20,edgecolor='black')
plt.title('Age Data Distribution')
plt.xlabel('Age Value')
plt.ylabel('Frequency')
```

```python
#Histogram
Sales_data =np.random.randint(100,5000,200)
print(Sales_data)
plt.figure(figsize=(6,4))

counts, bins, patches = plt.hist(Sales_data, bins=5, edgecolor='Red')
plt.title('Sales Data Distribution')
plt.xlabel('Sales Value')
plt.ylabel('Frequency')

for i in range(len(counts)):
    x = (bins[i] + bins[i+1]) / 2   # center of bin
    y = counts[i]
    plt.text(x, y, int(y), ha='center', va='bottom')

plt.show()
```
