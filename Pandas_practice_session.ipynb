```python
#Importing pandas
import pandas as pd
```

```python
# Series 1D array
s = pd.Series([100, 200, 300, 400])
print(s)
print(type(s))
```

```python
# 2D array

data = {'Name':["Arvind","Vinoth","Fathima","Jeni","Karthiga"],
        'Age': [25,30,26,28,29],
        'City':["Bangalore","Chennai","Mumbai","Chennai","Hydrabad"]}
print(data)
```

```python
df = pd.DataFrame(data)
df
```

```python
# Creating Dictionary
data = {
    'ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Age': [24, 30, 28, 35, 22, 40],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Berlin', 'Sydney'],
    'Salary': [50000, 60000, 55000, 70000, 48000, 75000],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'Marketing', 'Finance']
}

print(data)
```

```python
df = pd.DataFrame(data)
df
```

```python
#Accessing Columns (single element)

df['Age']
```

```python
#Accessing Columns (multiple element)

df[["Name","Age","City"]]
```

```python
#Accessing Rows (multiple rows)

df[0:3]
```

```python
#using loc to access elements(label based indexing)

print(df.loc[2,'Name'])
print(df.loc[1,'Name'])
```

```python
#using loc for slicing

df.loc[:2:2]
```

```python
#Accessing multiple element (using loc)

print(df.loc[2:3, ["Name", "Age","Salary"]])
```

```python
#Accessing element using iloc (Single row)

df.iloc[0]
```

```python
#Accessing single element using iloc(rows and columns)
df.iloc[0,3]
```

```python
#difference between loc and iloc

print(df.loc[0,"Name"])
print(df.iloc[0,1])
```

```python
#loc - includes the end lable
#iloc - excludes the end label
print("-------loc - OutPut------")
print(df.loc[0:2])
print("-------iloc - OutPut------")
print(df.iloc[0:2])
```

```python
#Filtering Data

df[df["Age"] <= 30]
```

```python
df[df["Salary"] >= 60000 ]
```

```python
df_dept = df[df["Department"] == "IT"]
print(df_dept)
```

```python
#Filter using logical operators

print("-------------And operator output-------")

df[(df["Age"] >= 30 ) & (df["Salary"] >= 60000)]



```

```python
print("-------------OR operator output-------")

df[(df["Age"] >= 35 ) | (df["Salary"] >= 50000)]
```

```python
#sorting Data based on  Age

df.sort_values("Age")
```

```python
#sort using salary

df.sort_values("Salary")
```

```python
#adding column

df.loc[6] = [107,"George",37,"Berlin",78000,"Marketing"]
df
```

```python
#adding column (len())

df.loc[len(df)] = [108,"Genelia",33,"Auckland",80000,"IT"]
df
```

```python
#updating value

df.loc[0,'Salary'] = 55000
df
```

```python
#update value(using loc) - do not know index number

df.loc[df['Name'] == "Genelia",'Age'] = 35
df
```

```python
#deleting Columns

df_new = df.drop("Department",axis = 1)
df_new
```
