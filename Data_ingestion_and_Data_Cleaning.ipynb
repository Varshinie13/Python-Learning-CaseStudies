Handlling null values(missing data)

```python
#returns null values in the columns

df.isnull().sum()
```

```python
#name column cleaning
df['Name'] = df['Name'].str.title()   #First letter caplitalise in the 'Name' column
df.head()
```

```python
#Age columns - use median when there is outliers

df['Age'].fillna(df['Age'].median(),inplace = True)

#df['Age'].fillna(20,inplace = True )               # fill 20 in Null places

#df['Age'].fillna(df['Age'].mean(),inplace = True)     #Age columns - you can use mean

#df['Age'].fillna(df['Age'].mode().[0],inplace = True)     #Age columns - you can use mode #if 2 modes - take 0 index value
```

```python
df.isnull().sum()
```

```python
#check the values in column

df['Department'].unique()
```

```python
#chance the department to upper case

df['Department'] = df['Department'].str.upper()
```

```python
#returns unique values of department column
df['Department'].unique()
```

```python
# chanege the Values
df['Department'].replace({'FINANACE':'Finance',
                          'FINANCE': 'Finance',
                          'SALES': 'Sales',
                          'MARKETING': 'Marketing'},inplace = True
                          )
```

```python
df
```

```python
#returns positions unique values
df['Position'].unique()
```

```python
#Remove the unknown character from position column

df['Position']=df['Position'].str.replace(r'[^a-zA-Z ]',' ', regex=True)
df
```

```python
#returns unique values in the columns
df['Salary'].unique()
```

```python
#Clean the salary

df['Salary']=df['Salary'].str.replace('k','000').str.replace(',','')
```

```python
#case conversion
df['Salary']=df['Salary'].astype(float)
```

```python
df['Salary'].describe()
```

```python
# Hierarchical imputation
# Level 1: Department + Position mean
dept_pos_mean = df.groupby(['Department','Position'])['Salary'].transform('median')

# # Level 2: Department mean
dept_mean = df.groupby('Department')['Salary'].transform('mean')

# # Level 3: Overall mean
overall_mean = df['Salary'].mean()

# Hierarchical imputation
df['Salary'] = (df['Salary']
    .fillna(dept_pos_mean)
    .fillna(dept_mean)
    .fillna(overall_mean)
)
```
