```python
#check the text is palindrome or not

text = input("Enter the text : ")
if text == text[::-1]:
  print("The given text is palindrome: ",text)
else:
  print("The given text is not palindrome: ",text)
```

```python
#check a number is prime or not

def is_prime(number):

    if number <= 1:
        return False


    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return False

    return True

try:
    user_num = int(input("Enter an integer to check: "))

    if is_prime(user_num):
        print(f"Yes, {user_num} is a prime number!")
    else:
        print(f"No, {user_num} is not a prime number.")

except ValueError:
    print("Invalid input. Please enter a whole number.")
```
