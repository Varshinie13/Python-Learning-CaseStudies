```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

Reading the web pages - wikipedia - The_World's_Billionaires

```python
url ="https://en.wikipedia.org/wiki/The_World%27s_Billionaires"
html = requests.get(url)
print(html.status_code)
print(html.text[:500])
```

HTML - Raw data from web page

```python
headers_agent = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/149.0.0.0 Safari/537.36"
}

url = "https://en.wikipedia.org/wiki/The_World%27s_Billionaires"
html = requests.get(url, headers=headers_agent).text
print(html)
```

Parse HTML Using BeautifulSoup

```python
soup = BeautifulSoup(html, "html.parser")    #get structured HTML tags
soup
```

```python
tables = soup.find_all("table", {"class": "wikitable"})
print(tables)
```

```python
tables = soup.find_all("table")           #get the class
for i, t in enumerate(tables):
    print( i, t.get("class"))
```

```python
tab = tables[2]       #call the table name to work on it
print(tab)
```

```python
headers = tab.find("tr").text # returns the row
print(headers)
```

```python
headers = tab.find('tr')
cols = [th.get_text(strip=True) for th in headers if th.get_text(strip=True)]
cols
```

```python
trs = tab.find_all('tr')[1:]
rows = []
for tr in trs:
    row = [td.get_text(strip=True) for td in tr.find_all(['td'])]
    rows.append(row)
print(rows)
```

```python
import pandas as pd
df = pd.DataFrame(rows,columns = cols)
df
```

```python
target_tables = [tables[i] for i in range(3, 10)]  # 2 to 6 inclusive
```

```python
year = 2025
new_dfs = []  # list to collect new tables

for idx, table in enumerate(target_tables, start=2):
    print(f"\n📘 Billionaires {year}")
    year -= 1

    # Extract column headers
    header_row = table.find("tr")
    cols = [th.get_text(strip=True) for th in header_row.find_all("th")]

    # Extract data rows
    rows = []
    for tr in table.find_all("tr")[1:]:
        row = [td.get_text(strip=True) for td in tr.find_all(["td"])]
        if row:
            rows.append(row)

    # Convert to DataFrame
    dfs = pd.DataFrame(rows, columns=cols if cols else None)
    new_dfs.append(dfs)

# Combine all new tables into a single DataFrame
new_data = pd.concat(new_dfs, ignore_index=True)

new_data
```

```python
# Append to existing df
df = pd.concat([df, new_data], ignore_index=True)
df
```

```python
#Save all tables to one excel file…
df.to_csv("billionaires.csv", index=False)
```
