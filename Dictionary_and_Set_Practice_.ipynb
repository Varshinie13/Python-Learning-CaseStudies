Dictionary Case Study

Student Data Management System Instructions:


```python
#Created Dictionary
Students = {"Name":"Vishnu","Age":24,"Course":"Python"}
print(Students)
```

```python
#Created Multiple Students Details

Students={
    "Name": ["Srikanth","Swetha","Vinitha","arjun"],
         "Age": [32,25,27,30],
         "Course": ["AI","Data Science","Data Analytics"]}

print(Students)
```

```python
# Adding a new student using Append

Students["Name"].append("Krishna")
Students["Age"].append(32)
Students["Course"].append("Java")
print(Students)
```

```python
#Display all student details using dictionary keys.

Students.items()
```

```python
#Access and Display
#First student's details using indexing

Students["Name"][0],Students["Age"][0],Students["Course"][0]
```

```python
#Newly added student's details using indexing

Students["Name"][-1],Students["Age"][-1],Students["Course"][-1]
```

Set - Case Study

Student Course Registration System using Set Instructions:

```python
#Create a set named courses with course names
Course_Name = {"Python","SQL","Excel","Power BI","Tableau","Statistics"}
Course_Name
```

```python
#Add course using add()

Course_Name.add("Java")
print(Course_Name)
```

```python
#Add multiple courses using update()

Course_Name.update(["R","C++","Pytorch"])
Course_Name
```

```python
#Remove course using remove() and discard()

Course_Name.remove("R")
print(Course_Name)

Course_Name.discard("C")
print(Course_Name)
```

```python
#Display all courses and total number of courses using len()

print(len(Course_Name))
print(Course_Name)
```
