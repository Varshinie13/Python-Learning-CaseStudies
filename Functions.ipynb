Write a function to find the largest of three numbers.

```python
def large_num():
  num1 = int(input("Enter first number: "))
  num2 = int(input("Enter second number: "))
  num3 = int(input("Enter third number: "))
  if num1 > num2 and num1 > num3:
    print(f"{num1} is the largest number")
  elif num2 > num1 and num2 > num3:
    print(f"{num2} is the largest number")
  else:
    print(f"{num3} is the largest number")
```

```python
large_num()
```

Write a function to calculate the factorial of a number using recursion.

```python
def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial(n-1)

factorial(5)
```

```python
factorial(7)
```

Create a function to check whether a string is a palindrome.

```python
def palindrome(user_str):
  if user_str == user_str[::-1]:
    print(f"{user_str} is a palindrome")
  else:
    print(f"{user_str} is not a palindrome")
palindrome("Name")
```

```python
palindrome("madam")
```

Write a function that accepts a list and returns the sum of all elements.

```python
def my_list(nums):
  print(f"The given list is : ",nums)
  return sum(nums)
result = my_list([10,20,30])
print(f"The sum of all elements in the list is : {result}")
```

Create a function using *args to find the average of multiple numbers.

```python
def average(*args):
  if len(args) < 2:
    print("Enter atleast 2 numbers")
  else:
    average = sum(args)/len(args)
    print("The average of the number is :",average)
average(20,30,40)
```

```python
average(10)
```

Write a function that accepts a list of student marks and uses a loop to count how many students scored above 75.

```python
def student_marks():
  print("Enter the student marks : ")
  student_mark = list(map(int,input().split()))
  print(f"The given list of student marks is : {student_mark}")
  count = 0
  for i in student_mark:
    if i > 75:
      count += 1

  print(f"Total number of students above 75 is : {count}")

student_marks()
```
