Write a function to find the largest of three numbers.

```python
def large_num():
  num1 = int(input("Enter first number: "))
  num2 = int(input("Enter second number: "))
  num3 = int(input("Enter third number: "))
  if num1 > num2 and num1 > num3:
    print(f"{num1} is the largest number")
  elif num2 > num1 and num2 > num3:
    print(f"{num2} is the largest number")
  else:
    print(f"{num3} is the largest number")
```

```python
large_num()
```

Write a function to calculate the factorial of a number using recursion.

```python
def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial(n-1)

factorial(5)
```

```python
factorial(7)
```

Create a function to check whether a string is a palindrome.

```python
def palindrome(user_str):
  if user_str == user_str[::-1]:
    print(f"{user_str} is a palindrome")
  else:
    print(f"{user_str} is not a palindrome")
palindrome("Name")
```

```python
palindrome("madam")
```

Create a function using *args to find the average of multiple numbers.

```python
def average(*args):
  if len(args) < 2:
    print("Enter atleast 2 numbers")
  else:
    average = sum(args)/len(args)
    print("The average of the number is :",average)
average(20,30,40)
```

```python
average(10)
```

Write a function that accepts a list of student marks and uses a loop to count how many students scored above 75.

```python
def student_marks():
  print("Enter the student marks : ")
  student_mark = list(map(int,input().split()))
  print(f"The given list of student marks is : {student_mark}")
  count = 0
  for i in student_mark:
    if i > 75:
      count += 1

  print(f"Total number of students above 75 is : {count}")

student_marks()
```

Create a function that takes a list of sales values and uses a loop to calculate the total sales greater than a given threshold.

```python
def Sales():
  print("Enter the Sales Value:  ")
  sales_amt = list(map(int,input().split()))
  print(f"The give Sales amt is : ",sales_amt)
  threshold = int(input("Enter the threshold value: "))
  total_sales = 0
  for i in sales_amt:
    if i > threshold:
      total_sales += i
  print(f"Total sales greater than the threshold is : {total_sales}")

Sales()
```

Create a function that takes a list of numbers and uses a loop to return only the even numbers in a new list.

```python
def numbers():
  print("Enter the numbers :  ")
  num = list(map(int,input().split()))
  print(f"The give numbers are : ",num)
  even_num = []
  for n in num:
    if n%2 == 0:
      even_num.append(n)
  if len(even_num) > 0:
    print(f"The even number in the list is : ",even_num)
  else:
    print("There is no even number in the list")

numbers()



```

Calculate Monthly Bonus for Employees:

Write a function that accepts a list of employee sales.

If sales are greater than 50,000, bonus is 10% of sales
If sales are between 30,000 and 50,000, bonus is 5%
Otherwise, no bonus

```python
def monthly_bonus_calc():
  print("Enter the sales amount of the employee: ")
  sales = list(map(int,input().split()))
  for s in sales:
    if s >50000:
      after_bonus = s+(s*0.1)
      print("After Bonus Salary : ", after_bonus)
    elif s>=30000 and s<=50000:
      after_bonus = s+(s*0.05)
      print("After Bonus Salary : ",after_bonus)
    else:
      print("No bonus. Salary remains",s)
monthly_bonus_calc()
```
