```python
#importing lib
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
df = pd.DataFrame({
    'Region': np.random.choice(['North', 'South', 'East', 'West'], 100),
    'Country': np.random.choice(['A', 'B', 'C', 'D', 'E'], 100),
    'Year': np.random.choice([2020, 2021, 2022, 2023, 2024], 100),
    'Population': np.random.randint(1_000_000, 100_000_000, 100),
    'GDP': np.random.randint(10_000, 100_000, 100),
    'UnemploymentRate': np.round(np.random.uniform(3, 15, 100), 2),
    'HappinessScore': np.round(np.random.uniform(0, 10, 100), 2),
    'IncomeGroup': np.random.choice(['Low', 'Middle', 'High'], 100)
})

# Adding GDP Outliers
df.loc[5, 'GDP'] = 200000
df.loc[15, 'GDP'] = 250000


# Adding Population Outliers
df.loc[10, 'Population'] = 500000000
df.loc[20, 'Population'] = 700000000
df.loc[30, 'Population'] = 900000000

sns.set(style="whitegrid", palette="Set2")

df
```

```python
# Bar Plot - Mean GDP by Region
sns.barplot(x='Region', y='GDP', data=df)

plt.title('Average GDP by Region')
plt.xlabel('Region')
plt.ylabel('Mean GDP')
plt.show()
```

```python
# Count Plot- Count of Income Groups
sns.countplot(x='IncomeGroup', data=df)
plt.title('Count of Records by Income Group')
plt.show()
```

```python
#Point Plot - Mean GDP by Region

sns.pointplot(x='Region', y='GDP', data=df)

plt.title('Mean GDP by Region')
plt.xlabel('Region')
plt.ylabel('Mean GDP')
plt.show()
```

```python
#BOX PLOT - GDP Distribution

sns.boxplot(y=df['GDP'])

plt.title('GDP Distribution')
plt.ylabel('GDP')
plt.show()
```

```python
# Box Plot -Unemployment Rate by Region
sns.boxplot(x='Region', y='UnemploymentRate', data=df)

plt.title('Unemployment Rate by Region')
plt.xlabel('Region')
plt.ylabel('Unemployment Rate')
plt.show()
```

```python
# Violin Plot Unemployment Rate by Region
sns.violinplot(x='Region', y='UnemploymentRate', data=df)

plt.title('Unemployment Rate by Region')
plt.xlabel('Region')
plt.ylabel('Unemployment Rate')
plt.show()
```

```python
#Boxen plot GDP Distribution by Region

sns.boxenplot(x='Region', y='GDP', data=df)

plt.title('GDP Distribution by Region')
plt.xlabel('Region')
plt.ylabel('GDP')
plt.show()
```

```python
# Strip Plot -Happiness Score by Income Group
sns.stripplot(x='IncomeGroup', y='HappinessScore', data=df)

plt.title('Happiness Score by Income Group')
plt.xlabel('Income Group')
plt.ylabel('Happiness Score')
plt.show()
```

```python
# Swarm Plot -Happiness Score by Income Group

sns.swarmplot(x='IncomeGroup', y='HappinessScore', data=df)

plt.title('Happiness Score by Income Group')
plt.xlabel('Income Group')
plt.ylabel('Happiness Score')
plt.show()
```

```python
# Scatter Plot -GDP vs Happiness Score
sns.scatterplot(x='GDP', y='HappinessScore', data=df)

plt.title('GDP vs Happiness Score')
plt.xlabel('GDP')
plt.ylabel('Happiness Score')
plt.show()
```

```python
# Line Plot - GDP Trend Over Years
plt.figure(figsize=(6,4))
sns.lineplot(x='Year', y='GDP', data=df)

plt.title('GDP Trend Over Years')
plt.xlabel('Year')
plt.ylabel('GDP')
plt.show()
```

```python
#Relplot -GDP vs Happiness Score
sns.relplot(x='GDP', y='HappinessScore',  kind='line',data=df)
plt.title('GDP Trend by Region')
plt.show()
```

```python
# relplot -Separate plot for each Income Group
sns.relplot(x='GDP', y='HappinessScore', col='IncomeGroup', data=df)
```

```python
# LM Plot-GDP vs Happiness Score with trend line
sns.lmplot(x='GDP', y='HappinessScore', data=df)
plt.suptitle('Regression by Income Group')
plt.show()
```

```python
# Histogram - GDP Distribution
sns.histplot(df['GDP'], bins=10,kde=True)
plt.title('GDP Distribution')
plt.show()
```

```python
# KDE Plot
sns.kdeplot(data=df, x='GDP',fill =True)
plt.title('GDP Density Distribution')
plt.xlabel('GDP')
plt.ylabel('Density')
plt.show()
```

```python
# Dist Plot (combined kde+ histogrm)
sns.displot(df['GDP'],kde =True,  bins=15)
plt.title('Distribution of Unemployment Rate')
plt.show()
```

```python
#ECDF Plot
sns.ecdfplot(data=df, x='HappinessScore')
plt.title('ECDF of Happiness Score')
plt.xlabel('Happiness Score')
plt.ylabel('Cumulative Probability')
plt.show()
```

```python
#rugplot-
sns.rugplot(data=df, x='HappinessScore')

plt.title('GDP KDE with Rug Plot')
plt.xlabel('GDP')
plt.show()
```

```python
#Correlation Heatmap
plt.figure(figsize=(6,4))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='Greens', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
df.corr(numeric_only=True)
```

```python
# Clustermap
sns.clustermap(df.corr(numeric_only=True), cmap='viridis', annot=True)
plt.suptitle('Clustered Correlation Heatmap')
plt.show()
```

```python
# Pair Plot
sns.pairplot(
    df[['GDP', 'Population', 'HappinessScore', 'UnemploymentRate']]
)
plt.suptitle('Pairwise Relationships')
plt.show()
```

```python
# Joint Plot
sns.jointplot(x='GDP', y='HappinessScore', data=df, kind='hex', cmap='coolwarm')
plt.suptitle('GDP vs Happiness (Joint Plot)')
plt.show()
```
